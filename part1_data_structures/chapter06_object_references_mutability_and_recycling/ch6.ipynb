### Identity, Equality, and Aliases

```python
charles = {"name": "Charles L. Dodgson", "born": 1832}
lewis = charles
lewis is charles

id(charles), id(lewis)

lewis["balance"] = 950

charles
```

```python
alex = {'name': 'Charles L. Dodgson', 'born': 1832, 'balance': 950}

alex is charles

alex == charles
```

### The Relative Immutability of Tuples

```python
t1 = (1,2,[30,40])
t2 = (1,2,[30,40])
t1 == t2

id(t1[-1])
t1[-1].append(99)
id(t1[-1])

t1 == t2
```
